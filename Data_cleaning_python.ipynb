```python
import pandas as pd
```

```python
df=pd.read_csv("orders.csv",na_values=['Not Available','unknown'],dtype={'Postal Code': str})
```

```python
df
```

```python
df['Ship Mode'].unique()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df['discount']=df['list_price']*df['discount_percent']*0.01
df['sales_price']=df['list_price']-df['discount']
df['profit']=df['sales_price']-df['cost_price']
```

```python
df['total_sales']=df['sales_price']*df['quantity']
df['total_profit']=df['profit']*df['quantity']
```

```python
df.dtypes
```

```python
df['order_date']=pd.to_datetime(df['order_date'],format="%Y-%m-%d")
```

```python
df.drop(columns=['list_price','cost_price','discount_percent'],inplace=True)
```

```python
import sqlalchemy as sal
import urllib.parse

# 1. Enter your MySQL credentials
user = "root"
password = urllib.parse.quote_plus("365PASS") # Replace with your MySQL password
host = "localhost"
port = "3306"
database = "orders"

# 2. Create the connection to MySQL
engine = sal.create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{database}")
conn = engine.connect()

# 3. Push your cleaned DataFrame (df) into the df_orders table
df.to_sql("df_orders", con=conn, index=False, if_exists="append")

print("Data loaded into MySQL df_orders successfully!")
```
